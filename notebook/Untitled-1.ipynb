```python
import pandas as pd
```

```python
train = pd.read_csv("/Users/clairedebadts/kaggle/customer-churn/data/train.csv")
test = pd.read_csv("/Users/clairedebadts/kaggle/customer-churn/data/test.csv")
```

```python
print("=== SHAPE ===")
print("Train:", train.shape)
print("Test:", test.shape)
```

```python
print("\n=== COLUMNS ===")
print(train.columns.tolist())
print(test.columns.tolist())
```

```python
print("\n=== DTYPES ===")
print(train.dtypes)
print(test.dtypes)
```

```python
print("\n=== HEAD ===")
print(train.head(5).to_string())
print(test.head(5).to_string())
```

```python
print("\n=== MISSING VALUES ===")
print(train.isnull().sum())
```

```python
print("\n=== TARGET DISTRIBUTION ===")
print(train["Churn"].value_counts(normalize=True))
```

```python
print("\n=== NUMERIC DESCRIBE ===")
print(train.describe().to_string())
print(test.describe().to_string())
```

```python
cat_cols = train.select_dtypes('object').columns.drop('Churn')

print("=== UNIQUE VALUES PER CATEGORICAL ===")
for col in cat_cols:
    print(f"\n{col}:")
    print(f"  train : {sorted(train[col].unique().tolist())}")
    print(f"  test  : {sorted(test[col].unique().tolist())}")
```
